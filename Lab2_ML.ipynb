# Manual Formulas Implementation

```python
data_brain = pd.read_csv('headbrain.csv')
df_brain = pd.DataFrame(data_brain)
df_brain
```

```python
X = df_brain['Head Size(cm^3)']
Y = df_brain['Brain Weight(grams)']
```

```python
X_Sum = np.sum(X)
X_Squared_Sum = np.sum(X*X)
N = len(X)
```

W1

```python
numerator = N*np.sum(X*Y)-(np.sum(X)*np.sum(Y))
denomerator = N*(np.sum(X*X))-(np.sum(X)**2)
W1 = numerator/denomerator
W1
```

W0


```python
W0 = (np.sum(Y)-W1*(np.sum(X)))/N
W0
```

```python
x_max = np.max(X)
x_min = np.min(X)
x = np.linspace(x_min,x_max)
```

```python
y = W0 + (W1*x)
```

#Plotting

```python
plt.scatter(X,Y,color='Blue')
plt.plot(x,y,color='Red')
plt.show()
```

RMSE

```python
Rmse = 0
for i in range(N):
  y_predict = W0+(W1*X[i])
  Rmse += (Y[i]-y_predict)**2
RMSE = np.sqrt(Rmse/N)
RMSE
```

# R2 Score

```python
Y_Mean = np.mean(Y)
for i in range(N):
  y_predict1 = W0+(W1*X[i])
  nume = (Y[i]-y_predict1)**2
  deno = (Y[i]-Y_Mean)**2

R2 = 1- (nume/deno)
R2
```

# Activity 1

```python
data_med = pd.read_csv('/content/Medical Cost Personal Datasets.csv')
df_med = pd.DataFrame(data_med)
df_med
```

```python
X_Med = df_med.drop(['sex','charges'],axis = 1)
Y_Med = df_med['charges']
```

```python
df_med['smoker'] = df_med['smoker'].map({'no':0,'yes':1})
```

```python
df_med = pd.get_dummies(df_med,columns=['region'],dtype=int)
```

```python
df_med
```

# Scaling Features

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_med[['age','bmi']] = scaler.fit_transform(df_med[['age','bmi']])
```

```python
df_med
```

# Train Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error,r2_score
LR1 = LinearRegression()
x_train,x_test,y_train,y_test = train_test_split(X_Med,Y_Med,test_size=0.2,random_state=42)
LR1.fit(x_train,y_train)
Prediction = LR1.predict(x_test)
RMSE = np.sqrt(root_mean_squared_error(y_test,Prediction))
R2_Score = r2_score(y_test,Prediction)
print('RMSE',RMSE)
print('R2_Score',R2_Score)



```

# Plotting

```python
plt.scatter(y_test,Prediction,color='Blue')
plt.plot([y_test.min(),y_test.max()],[y_test.min(),y_test.max()],color='Red')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.show()
```

```python
data_churn = pd.read_csv('/content/Telco Customer Churn.csv')
df_churn = pd.DataFrame(data_churn)
df_churn
```

```python
# df_churn = pd.get_dummies(df_churn,columns=['Contract'],dtype=int)
df_churn = pd.get_dummies(df_churn,columns=['InternetService'],dtype=int)

df_churn
```

```python
df_churn['Churn'] = df_churn['Churn'].map({'No':0,'Yes':1})
```

```python
df_churn
```

# Scaling


```python
# Tenure,MonthlyCharges,Contract,Iternetservices
df_churn[['tenure','MonthlyCharges']]  = scaler.fit_transform(df_churn[['tenure','MonthlyCharges']])
```

```python
df_churn
```

# Train Model

```python
df_churn.columns
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
LRR = LogisticRegression()
X_Churn = df_churn[['tenure','InternetService_DSL','InternetService_Fiber optic','InternetService_No','MonthlyCharges','Contract_Month-to-month','Contract_One year', 'Contract_Two year']]
Y_churn = df_churn['Churn']


xc_train,xc_test,yc_train,yc_test = train_test_split(X_Churn,Y_churn,test_size=0.2,random_state=42)
LRR.fit(xc_train,yc_train)
predict = LRR.predict(xc_test)
acc = accuracy_score(yc_test,predict)
prec = precision_score(yc_test,predict)
recal = recall_score(yc_test,predict)
f1 = f1_score(yc_test,predict)

print('Accuracy',acc)
print('Precision',prec)
print('Recall',recal)
print('f1',f1)





```
